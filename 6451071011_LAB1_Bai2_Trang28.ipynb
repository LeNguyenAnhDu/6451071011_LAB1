## Câu 2. Liệt kê 5 thuộc tính có nhiều giá trị thiếu nhất

Ngoài giá trị rỗng thật sự, kiểm tra riêng nhãn `unknown` vì bộ dữ liệu dùng nhãn này cho một số thông tin chưa biết.

## Câu 3. Khảo sát thuộc tính với biến quyết định `y`

Tệp không có giới tính. Dùng tuổi và tình trạng hôn nhân để so sánh nhóm đăng ký/không đăng ký; khảo sát thêm thời lượng contact cuối (`duration`).


```python
fig,ax=plt.subplots(figsize=(9,5))
sns.boxplot(data=df,x="y",y="age",hue="marital",showfliers=False,ax=ax)
ax.set(title="Tuổi theo tình trạng hôn nhân và đăng ký tiền gửi",xlabel="Đăng ký (y)",ylabel="Tuổi")
plt.tight_layout()
print(f"Biểu đồ tuổi × tình trạng hôn nhân × y | {STUDENT_INFO}"); plt.show()
fig,ax=plt.subplots(figsize=(8,5))
sns.boxplot(data=df,x="y",y="duration",showfliers=False,ax=ax,color="#4C78A8")
ax.set(title="Thời lượng contact cuối theo kết quả đăng ký",xlabel="Đăng ký (y)",ylabel="Thời lượng (giây)")
plt.tight_layout()
print(f"Biểu đồ duration theo y | {STUDENT_INFO}"); plt.show()
rates=df.groupby(["marital","y"],observed=False).size().unstack(fill_value=0)
rates=rates.div(rates.sum(axis=1),axis=0).mul(100).round(1)
print(f"Tỷ lệ y theo tình trạng hôn nhân (%) | {STUDENT_INFO}"); display(rates)
print("Nhận xét: so sánh tỷ lệ trong bảng và phân bố trên biểu đồ. Duration thường dài hơn ở nhóm đăng ký, nhưng chỉ được biết sau cuộc gọi nên không dùng làm biến đầu vào cho nhắm mục tiêu trước khi gọi.")
```

## Câu 4. Học vấn, tình trạng hôn nhân, khoản vay và khả năng đăng ký

So sánh tỷ lệ `y=yes` theo học vấn, phân tầng theo tình trạng hôn nhân và khoản vay cá nhân.


```python
q4=df.groupby(["marital","education","loan"],observed=False).y.apply(lambda s:s.eq("yes").mean()*100).round(1).rename("Tỷ lệ đăng ký (%)").reset_index()
print(f"Tỷ lệ đăng ký theo marital, education, loan | {STUDENT_INFO}"); display(q4)
fig,axs=plt.subplots(1,3,figsize=(15,5),sharey=True)
for ax,marital in zip(axs,["married","single","divorced"]):
 part=q4[q4.marital==marital]
 sns.barplot(data=part,x="education",y="Tỷ lệ đăng ký (%)",hue="loan",order=["primary","secondary","tertiary","unknown"],ax=ax)
 ax.set(title=marital.title(),xlabel="Trình độ học vấn",ylabel="Tỷ lệ đăng ký (%)" if marital=="married" else "")
 ax.tick_params(axis="x",rotation=25); ax.set_ylim(0,100)
fig.suptitle("Tỷ lệ đăng ký theo học vấn, tình trạng hôn nhân và khoản vay")
plt.tight_layout()
print(f"Biểu đồ phân tầng theo marital và loan | {STUDENT_INFO}"); plt.show()
print("Nhận xét: các mức tỷ lệ có thể khác nhau theo học vấn, khoản vay và tình trạng hôn nhân; đọc theo từng nhóm trên biểu đồ, không diễn giải chênh lệch mô tả thành quan hệ nhân quả.")
```

## Câu 5. Số lần liên hệ và khả năng đăng ký

`campaign` là số lần liên hệ trong chiến dịch hiện tại; `previous` là số lần liên hệ trước chiến dịch này. Tổng hai biến dùng làm chỉ báo số lần liên hệ, không phải số thành viên gia đình.


```python
q5=df.copy()
q5["total_contacts"]=q5["campaign"]+q5["previous"]
q5["contact_group"]=pd.cut(q5["total_contacts"],bins=[-1,1,3,np.inf],labels=["1 lần","2–3 lần","4 lần trở lên"])
counts=q5["total_contacts"].value_counts().sort_index().rename_axis("Tổng lần liên hệ").to_frame("Số khách hàng")
rates=q5.groupby("contact_group",observed=False).y.apply(lambda s:s.eq("yes").mean()*100).round(1).rename("Tỷ lệ đăng ký (%)").to_frame()
print(f"Số khách hàng theo tổng số lần liên hệ | {STUDENT_INFO}"); display(counts)
print(f"Tỷ lệ đăng ký theo nhóm liên hệ | {STUDENT_INFO}"); display(rates)
fig,ax=plt.subplots(figsize=(8,5))
sns.barplot(data=q5,x="contact_group",y=q5.y.eq("yes").astype(int),order=["1 lần","2–3 lần","4 lần trở lên"],color="#59A14F",ax=ax)
ax.set(title="Số lần liên hệ và tỷ lệ đăng ký",xlabel="Tổng số lần liên hệ",ylabel="Tỷ lệ đăng ký"); ax.set_ylim(0,1); plt.tight_layout()
print(f"Biểu đồ số lần liên hệ × y | {STUDENT_INFO}"); plt.show()
print("Nhận xét: đối chiếu tỷ lệ theo nhóm để quan sát mối liên hệ; liên hệ nhiều lần không đảm bảo khả năng đăng ký cao hơn.")
```

## Câu 6. Tiền xử lý giá trị thiếu và loại thuộc tính không phù hợp

Các ô trống thật sự được điền bằng trung vị (biến số) hoặc mode (biến phân loại). Nhãn `unknown` được giữ thành một mức riêng để không làm mất thông tin chưa biết. Tạo bảng dữ liệu dùng cho phân tích trước chiến dịch và bỏ `duration`: đây là thông tin chỉ có sau cuộc gọi cuối, không phù hợp nếu mục tiêu là nhắm khách trước khi gọi. Đã khảo sát `duration` ở Câu 3.


```python
clean_df=df.copy()
num_cols=clean_df.select_dtypes(include="number").columns
cat_cols=clean_df.select_dtypes(exclude="number").columns
for col in num_cols:
 clean_df[col]=clean_df[col].fillna(clean_df[col].median())
for col in cat_cols:
 if clean_df[col].isna().any(): clean_df[col]=clean_df[col].fillna(clean_df[col].mode().iloc[0])
clean_df=clean_df.drop(columns=["duration"],errors="ignore")
print(f"Kích thước sau tiền xử lý: {clean_df.shape} | {STUDENT_INFO}")
print(f"Số NA còn lại theo thuộc tính | {STUDENT_INFO}"); display(clean_df.isna().sum().to_frame("Số giá trị rỗng"))
print(f"Các cột còn lại | {STUDENT_INFO}"); display(pd.DataFrame({"Thuộc tính":clean_df.columns}))
```

## Câu 7. Mã hóa thuộc tính và chia nhóm

Mã hóa nhị phân yes/no thành 1/0; biến phân loại danh nghĩa bằng one-hot encoding. Chia tuổi thành 8 nhóm và số dư tài khoản thành 6 nhóm gần cân bằng bằng phân vị (`qcut`); in số mẫu để kiểm tra.


```python
clean_df["age_group"]=pd.qcut(clean_df["age"],q=8,labels=False,duplicates="drop")+1
clean_df["balance_group"]=pd.qcut(clean_df["balance"],q=6,labels=False,duplicates="drop")+1
binary_cols=["default","housing","loan","y"]
for col in binary_cols: clean_df[col]=clean_df[col].map({"no":0,"yes":1}).astype("int64")
group_counts=pd.concat([clean_df.age_group.value_counts().sort_index().rename("Số mẫu nhóm tuổi"),clean_df.balance_group.value_counts().sort_index().rename("Số mẫu nhóm số dư")],axis=1).fillna(0).astype(int)
print(f"Số lượng mẫu trong các nhóm phân vị | {STUDENT_INFO}"); display(group_counts)
cat_cols=["job","marital","education","contact","month","poutcome"]
encoded_df=pd.get_dummies(clean_df,columns=cat_cols,dtype=int)
print(f"Bản ghi sau mã hóa (5 dòng đầu, một số cột) | {STUDENT_INFO}")
display(encoded_df.iloc[:5,:14])
```

## Câu 8. Tạo thuộc tính mới

Tạo tổng số lần liên hệ và thời lượng trung bình trên mỗi lần liên hệ có làm trơn `+1`; tạo thêm tương tác tuổi × số lần liên hệ trong chiến dịch.


```python
clean_df["total_contacts"]=clean_df["campaign"]+clean_df["previous"]
clean_df["age_campaign"]=clean_df["age"]*clean_df["campaign"]
clean_df["duration_per_contact"]=df["duration"]/(clean_df["total_contacts"]+1)
print(f"Các thuộc tính mới (5 dòng đầu) | {STUDENT_INFO}")
display(clean_df[["age","campaign","previous","total_contacts","age_campaign","duration_per_contact","y"]].head())
print(f"Thống kê thuộc tính mới | {STUDENT_INFO}")
display(clean_df[["total_contacts","age_campaign","duration_per_contact"]].describe().T)
```
